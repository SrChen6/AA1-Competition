# Clustering over all train data (no categorical data)

We wish to evaluate how our learning data is separated when we use no categorical data. To better visualize the dataset, we are going to apply PCA first.

```python
num_data = X_learn.drop(columns = ["Female"])

corr = num_data.corr()
corr.style.background_gradient(cmap='RdBu')
```

```python
eigs, U = np.linalg.eig(corr) 
X_pca = num_data @ U

print(eigs)
print(X_pca.shape)
```

```python
x = X_pca.loc[:, 0]
y = X_pca.loc[:, 1]

plt.figure(figsize = (6, 4))
_ = plt.scatter(x = x, y = y, c = learn_data['Target'], cmap = "RdBu")
```

The data cannot be correctly separated with just the first two dimensions, so we will be using all the dimensions of PCA to fit clusters.

We fit from 2 to 10 clusters to our data and evaluate the CH, silhouette and DB metrics.

```python
index = pd.MultiIndex.from_arrays([['kmeans'], [2]], names=('model', 'K'))

results_df = pd.DataFrame(index=index, columns= ['CH score', 'Silhouette score', 'DB score'])
```

```python
def compute_clustering_and_plot(K, data, results):

    # Train the model with k-means with given K
    kmeans = KMeans(n_clusters = K, max_iter=100)
    kmeans.fit(data)
    data['kmeans_labels'] = kmeans.labels_

    # Plot results
    sns.scatterplot(x = "0", y = "1", hue='kmeans_labels', data = data, palette = "RdBu")
    plt.plot(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:, 1], 'bo', markersize=12, alpha=0.7)
    plt.xlim(plt.ylim())
    plt.title(f'K Means predictions with K={K}')
    plt.show()

    # Compute metrics
    CH = calinski_harabasz_score(data, kmeans.labels_ )
    S = silhouette_score(data, kmeans.labels_ )
    DB = davies_bouldin_score(data, kmeans.labels_ )
    
    # store metrics
    results.loc[('kmeans', k),:] = [CH,S,DB]
    return results

X_pca.columns = X_pca.columns.astype(str)
K_values = [2, 3, 5, 7, 10]
for k in K_values:
    results = compute_clustering_and_plot(k, X_pca, results_df)
```

```python
results
```

The best clustering is given when using 2 or 10 clusters. We are going to consider 10 clusters from now on. We can check the probabilities of being sick / healthy in each cluster.

The clusters are good enough at finding out which patients are sick and which are healthy... the expected accuracy for this classifier is 80%. Of course, this is assuming that the clusters we have made are accurate enough and are similar to the test data.

```python
probtable = pd.DataFrame(columns = ["0", "1"])

K = 10
for i in range(K):
    total = sum(X_pca.kmeans_labels == i)
    prob0 = sum(learn_data["Target"][X_pca.kmeans_labels == i] == 0) / total
    prob1 = sum(learn_data["Target"][X_pca.kmeans_labels == i] == 1) / total
    probtable.loc[i,:] = [prob0, prob1]

probtable
```

```python
print("Expected accuracy:", np.mean(np.max(probtable, axis = 1)))
```

## Gaussian Mixtures

We can try to fit Gaussian Mixtures in an attempt to improve this result... Using basic diagonal matrices does not give us good results.

```python
gm = GaussianMixture(n_components = 10, covariance_type = 'diag').fit(X_pca.loc[:,:])
print('\nLOG Likelihood=', gm.lower_bound_)
print('\nWEIGHTS=')
pd.DataFrame(gm.weights_)
print('\nMEANS=')
pd.DataFrame(gm.means_)
print('\nCOV=')
pd.DataFrame(gm.covariances_)
```

```python
X_pca['gm_labels'] = gm.predict(X_pca.loc[:,:])
# Plot results
sns.scatterplot(x = '0', y = '1', hue = 'gm_labels', data = X_pca, palette = "RdBu")
```

```python
probtable = pd.DataFrame(columns = ["0", "1"])

K = 10
for i in range(K):
    total = sum(X_pca.gm_labels == i)
    prob0 = sum(learn_data["Target"][X_pca.gm_labels == i] == 0) / total
    prob1 = sum(learn_data["Target"][X_pca.gm_labels == i] == 1) / total
    probtable.loc[i,:] = [prob0, prob1]

probtable
```

```python
print("Expected accuracy:", np.mean(np.max(probtable, axis = 1)))
```

If we attempt to use full covariance matrices, we obtain similar accuracy to the K-Means method.

```python
gm = GaussianMixture(n_components = 10, covariance_type='full').fit(X_pca.loc[:,[str(n) for n in range(10)]])
print('\nLOG Likelihood=', gm.lower_bound_)
print('\nWEIGHTS=')
pd.DataFrame(gm.weights_)
print('\nMEANS=')
pd.DataFrame(gm.means_)
print('\nCOV=')
gm.covariances_
```

```python
X_pca['gm_labels'] = gm.predict(X_pca.loc[:,[str(n) for n in range(10)]])
# Plot results
sns.scatterplot(x = '0', y = '1', hue = 'gm_labels', data = X_pca, palette = "RdBu")
```

```python
print("Expected accuracy:", np.mean(np.max(probtable, axis = 1)))
```
