Now we will take a look at variables separately, with the use of histograms and boxplots.

```python
def histbox (data, column: str, n_bins: int = 10) -> None:
    fig, axes = plt.subplots(1, 2, gridspec_kw = {'width_ratios': [1, 4]}, figsize=(9,4))
    sns.boxplot(data = data, x = 'Target', y = column, hue = 'Target', ax = axes[0])
    
    sick = data[data["Target"] == 0]
    healthy = data[data["Target"] == 1]
    mean_sick = np.mean(sick[column])
    mean_healthy = np.mean(healthy[column])
    std_sick = np.std(sick[column])
    std_healthy = np.std(healthy[column])
    prop_sick = len(sick) / len(data)
    prop_healthy = len(healthy) / len(data)
    
    x = np.linspace(np.min(sick[column]), np.max(sick[column]), num = 500)
    y = np.linspace(np.min(healthy[column]), np.max(healthy[column]), num = 500)
    ax = sns.histplot(data = data, x = column, hue = 'Target', ax = axes[1],
                      bins = n_bins, stat = "density")
    ax.plot(x, norm.pdf(x, loc = mean_sick, scale = std_sick) * prop_sick)
    ax.plot(y, norm.pdf(y, loc = mean_healthy, scale = std_healthy) * prop_healthy)
```

## Experimental: should ages be categorical?

Unlike the other variables, which are clearly numerical or categorical, it is difficult to know whether age is better off as a categorical variable or a numerical one.

To better understand if treating age as a categorical variable is better for prediction, let us analyze the differences between different ranges of age in distribution of our data.

```python
data_v4 = data_v3.copy()
data_v4.index = range(0, data_v4.shape[0])
data_v4.head()
```

```python
ranges = [(0, 15), (16, 30), (31, 45), (46, 60), (60, 75), (75, 90)]

for i in range(data_v4.shape[0]):
    for x, y in ranges:
        if x <= data_v4.Age[i] <= y:
            data_v4.Age[i] = f"{x}-{y}"
            break

data_v4.head()
data_v4.Age.value_counts()
```

This distribution we have chosen is not well distributed because the young group and the elder group have very few observations. So we are going to change the classes to have a better distribution.

```python
data_v4 = data_v3.copy()
data_v4.index = range(0, data_v4.shape[0])

ranges = [(0, 30), (31, 45), (46, 60), (61, 90)]

for i in range(data_v4.shape[0]):
    for x, y in ranges:
        if x <= data_v4.Age[i] <= y:
            data_v4.Age[i] = f"{x}-{y}"
            break

pd.crosstab(data_v4.Age,
            data_v4.Target, 
            margins = False)
```

```python
def plot_PCA (X: np.ndarray, target: np.ndarray) -> np.ndarray:
    corr = X.corr()

    eigs, U = np.linalg.eig(corr) 
    X_pca = X @ U

    print(eigs)
    print(X_pca.shape)

    x = X_pca.loc[:, 0]
    y = X_pca.loc[:, 1]
    z = X_pca.loc[:, 2]

    plt.figure(figsize = (6, 4))
    _ = plt.scatter(x = x, y = y, c = target, cmap = "RdBu")
    _ = plt.grid()
    
    fig = plt.figure(figsize = (7, 8))
    ax = fig.add_subplot(projection = '3d')
    ax.scatter(xs = x, ys = y, zs = z, c = target, cmap = "RdBu")

    _ = ax.set_xlabel("2nd dim")
    _ = ax.set_ylabel("1st dim")
    _ = ax.set_zlabel("3rd dim")
    return U
```

```python
numdata_v4 = data_v4.drop(columns = ["Age", "Female", "Target"])

X_0_30 = numdata_v4[data_v4.Age == "0-30"]
y_0_30 = data_v4.Target[data_v4.Age == "0-30"]
plot_PCA(X_0_30, y_0_30)
```

```python
X_31_45 = numdata_v4[data_v4.Age == "31-45"]
y_31_45 = data_v4.Target[data_v4.Age == "31-45"]
plot_PCA(X_31_45, y_31_45)
```

```python
X_46_60 = numdata_v4[data_v4.Age == "46-60"]
y_46_60 = data_v4.Target[data_v4.Age == "46-60"]
plot_PCA(X_46_60, y_46_60)
```

```python
X_61_90 = numdata_v4[data_v4.Age == "61-90"]
y_61_90 = data_v4.Target[data_v4.Age == "61-90"]
plot_PCA(X_61_90, y_61_90)
```

## Experimental: reducing dimension

We may benefit from reducing dimension. We have a lot of dimensions in PCA which hold very small amounts of variability, but they might still be relevant. We can try to reduce dimensionality by applying ANOVA on the principal component and checking the most important components for predicting the target variable.

```python
data_v5 = data_v3.copy()
data_v5_num = data_v3.drop(columns = ["Female", "Target"])
U = plot_PCA(data_v5_num, data_v5.Target)
X_pca = data_v5_num @ U
```

We can also obtain an approximated information matrix to gauge how much information about the target variable we obtain from the principal components. We

```python
from sklearn.metrics import mutual_info_score

def calc_MI(x, target, bins):
    minv = np.min(x)
    maxv = np.max(x)
    binsize = (maxv - minv) / bins
    x_ = np.array(x.copy())
    for i in range(x_.shape[0]):
        for j in range(bins):
            if minv + j * binsize <= x_[i] <= minv + (j + 1) * binsize:
                x_[i] = j
                break
    return mutual_info_score(x_, target)

Imatt = np.ndarray(shape = (11))
bins = 15
for i in range(10):
    Imatt[i] = calc_MI(data_v5_num.iloc[:, i], data_v5.Target, bins)
Imatt[10] = mutual_info_score(data_v5.Female, data_v5.Target)
Imatt = Imatt.reshape((1, -1))

columns = ['Age', 'TB', 'DB', 'Alkphos', 'Sgpt', 'Sgot', 'TP', 'ALB',
       'AR', 'BilRatio', 'Female']
Imatt = pd.DataFrame(data = Imatt, columns = columns)
sns.heatmap(Imatt, annot=True)
```
